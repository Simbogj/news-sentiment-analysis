# Task 1: Exploratory Data Analysis for Financial News Sentiment

This notebook establishes the initial EDA pipeline for the financial news dataset and prepares the project for Task 1 deliverables. It covers headline statistics, publisher analysis, publication timing, and keyword/topic extraction.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

import warnings
warnings.filterwarnings("ignore")

# Convert 'date' column to datetime
# df['date'] = pd.to_datetime(df['date'])
```

## 1. Dataset load

Review the schema and basic record counts for the news dataset.

```python
#Dataset Load Dataset
df = pd.read_csv("../data/raw/raw_analyst_ratings.csv")
df.head()
```

```python
#Date conversion
df['date'] = pd.to_datetime(df['date'], errors='coerce')
```

## 2. Headline Length Analysis

Analyze the distribution of headline character counts and word counts.

```python
# headline length analysis
df['headline_length'] = df['headline'].apply(len)
df['headline_length'].describe()
plt.figure(figsize=(10,5))
sns.histplot(df['headline_length'], bins=30)
plt.title("Headline Length Distribution")
plt.show()
```

## 3. Publisher Analysis

Identify the most active publishers and compare their output volume.

```python
publisher_counts = df['publisher'].value_counts().head(10)
publisher_counts

plt.figure(figsize=(12,6))
publisher_counts.plot(kind='bar')
plt.title("Top Publishers")
plt.show()
```

## 4. News Volume Over Time
Create time series summaries of daily news.

```python
daily_news = df.groupby(df['date'].dt.date).size()


plt.figure(figsize=(14,6))
daily_news.plot()
plt.title("Daily News Volume")
plt.xlabel("Date")
plt.ylabel("Number of Articles")
plt.show()
```

## 5. Analyze Publishing Hours


```python
# Extract hour from date
df['hour'] = df['date'].dt.hour

#Count by hour 
hourly_news = df['hour'].value_counts().sort_index()

plt.figure(figsize=(12,5))
hourly_news.plot(kind='bar')
plt.title("News Publishing Hours")
plt.xlabel("Hour")
plt.ylabel("Article Count")
plt.show()
```

## 6. Keyword Analysis 

Extract the most frequent terms and n-grams in headlines using TF-IDF and count-based methods.

```python
tfidf = TfidfVectorizer(
    stop_words='english',
    max_features=20
)

X = tfidf.fit_transform(df['headline'])

keywords = tfidf.get_feature_names_out()
keywords
```

## 7. Visualize Top Keywords

```python
scores = X.sum(axis=0).A1

keyword_df = pd.DataFrame({
    'keyword': keywords,
    'score': scores
})

keyword_df = keyword_df.sort_values(by='score', ascending=False)

plt.figure(figsize=(12,6))
sns.barplot(
    data=keyword_df,
    x='score',
    y='keyword'
)
plt.title("Top Keywords")
plt.show()
```
